# Cross-patient semantic-decoding time-course panels

Paper figure group **semantic_regression** (GloVe), for **both naming tasks**:
picture naming (12 participants) and auditory naming (6 participants).

Four metrics per task: independent-centroid balanced category accuracy, and raw word-retrieval
top-1 / top-3 / top-5 accuracy (the three word panels share a y-scale). Each panel overlays all
participants of that task (distinct colours, fixed per participant), with a per-participant
significance raster below the chance line (per-bin one-sided permutation test, obs mean vs. the
99th percentile of the shuffled null), cue markers (mean ± 1 s.d. across participants), and an
x-axis in seconds with 0 at that task's alignment cue (trial onset / auditory stimulus onset).
The cross-task figure `11_combined_both_tasks` puts the two tasks side by side (rows = metric,
columns = task) on a shared per-family y-scale.

All plotting/statistics live in `semantic_regression_panels.py` (same folder). This notebook just
drives it and displays the saved PNGs. PDFs are written with `pdf.fonttype = 42` (editable text).

Source data → `source_data/` (`source_data.csv`, `cue_timing.csv`, `peak_rise_stats.csv`; each has
a `task` column). Per-epoch obs/null arrays are cached per task in `panels_cache_{task}_GloVe.npz`;
pass e.g. `rebuild_cache=['auditory']` to rebuild one from the result PKLs (needs the Speech env).
